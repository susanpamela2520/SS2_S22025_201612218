```python
import pandas as pd
from sqlalchemy import create_engine
import pymysql

pd.set_option('display.max_columns', None)
```

# Conexion a la DB

```python
USUARIO = 'root'          
PASSWORD = 'root'  
HOST = 'localhost'         
PORT = '3306'             
DB_NAME = 'semi2'    

cadena_conexion = f"mysql+pymysql://{USUARIO}:{PASSWORD}@{HOST}:{PORT}/{DB_NAME}"

try:
    engine = create_engine(cadena_conexion)
    with engine.connect() as conn:
        print(f"✅ Conexión exitosa a la base de datos MySQL: {DB_NAME}")
except Exception as e:
    print("❌ Error al conectar a MySQL. Verifica tus credenciales.")
    print(f"Detalle: {e}")
    exit()
```

# Carga de datos desde archivos .csv

```python
# Carga de CSVs
try:
    df_compras = pd.read_csv('./archivosCarga/compras.csv', encoding='utf-8')
    df_ventas = pd.read_csv('./archivosCarga/ventas.csv', encoding='utf-8')
    print("Archivos cargados correctamente.")
except FileNotFoundError:
    print("Error: Sube los archivos compras.csv y ventas.csv al mismo directorio del notebook.")

# Muestra las primeras filas para verificar
display(df_compras.head(10))
display(df_ventas.head(10))
```

# Transformación y limpieza 

```python
# Función de limpieza para fechas
def limpiar_fecha(fecha_str):
    if pd.isna(fecha_str): return None
    # Corrige errores comunes de OCR (Z por 2, O por 0)
    fecha_str = str(fecha_str).replace('Z', '2').replace('O', '0')
    return fecha_str

# Limpieza de Fechas 
df_compras['Fecha'] = pd.to_datetime(df_compras['Fecha'].apply(limpiar_fecha), dayfirst=True, errors='coerce')
df_ventas['Fecha'] = pd.to_datetime(df_ventas['Fecha'].apply(limpiar_fecha), dayfirst=True, errors='coerce')

# Eliminar filas que no tengan fecha válida
df_compras.dropna(subset=['Fecha'], inplace=True)
df_ventas.dropna(subset=['Fecha'], inplace=True)

# Limpieza de Valores Negativos
df_ventas['PrecioUnitario'] = df_ventas['PrecioUnitario'].abs()
df_ventas['Unidades'] = df_ventas['Unidades'].abs()

df_compras['CostoUnitario'] = df_compras['CostoUnitario'].abs()
df_compras['Unidades'] = df_compras['Unidades'].abs()

print("Transformación y limpieza completada.")

display(df_compras.head(3))
display(df_ventas.head(3))
```

# Creacion de dimensiones y hechos

```python
# Dimensión Fecha
fechas_unicas = pd.concat([df_compras['Fecha'], df_ventas['Fecha']]).unique()
dim_fecha = pd.DataFrame({'Fecha': fechas_unicas})
dim_fecha['id_fecha'] = dim_fecha['Fecha'].dt.strftime('%Y%m%d').astype(int)
dim_fecha['anio'] = dim_fecha['Fecha'].dt.year
dim_fecha['mes'] = dim_fecha['Fecha'].dt.month
nombres_meses = {1:'Enero', 2:'Febrero', 3:'Marzo', 4:'Abril', 5:'Mayo', 6:'Junio', 
                 7:'Julio', 8:'Agosto', 9:'Septiembre', 10:'Octubre', 11:'Noviembre', 12:'Diciembre'}
dim_fecha['nombre_mes'] = dim_fecha['mes'].map(nombres_meses)
dim_fecha['dia'] = dim_fecha['Fecha'].dt.day
dim_fecha['trimestre'] = dim_fecha['Fecha'].dt.quarter.apply(lambda x: f"Q{x}")

dim_fecha_final = dim_fecha[['id_fecha', 'anio', 'nombre_mes', 'trimestre', 'dia', 'mes']].copy()

# Dimensión Producto (Unificando compras y ventas)
cols_prod = ['CodProducto', 'NombreProducto', 'MarcaProducto', 'Categoria']
dim_producto = pd.concat([df_compras[cols_prod], df_ventas[cols_prod]]).drop_duplicates(subset=['CodProducto']).reset_index(drop=True)

# Dimensión Sucursal
cols_suc = ['CodSucursal', 'NombreSucursal', 'Region', 'Departamento']
dim_sucursal = pd.concat([df_compras[cols_suc], df_ventas[cols_suc]]).drop_duplicates(subset=['CodSucursal']).reset_index(drop=True)

# Otras dimensiones
dim_cliente = df_ventas[['CodCliente', 'NombreCliente', 'TipoCliente']].drop_duplicates(subset=['CodCliente']).dropna(subset=['CodCliente'])
dim_vendedor = df_ventas[['CodVendedor', 'NombreVendedor']].drop_duplicates(subset=['CodVendedor']).dropna(subset=['CodVendedor'])
dim_proveedor = df_compras[['CodProveedor', 'NombreProveedor']].drop_duplicates(subset=['CodProveedor']).dropna(subset=['CodProveedor'])


print("Dimensiones creadas.")
display(dim_producto.head(3))

mapa_fechas = dim_fecha.set_index('Fecha')['id_fecha'].to_dict()


# -- Fac Compras --
fac_compras = df_compras.copy()
fac_compras['id_fecha'] = fac_compras['Fecha'].map(mapa_fechas)
fac_compras['total_compra'] = fac_compras['Unidades'] * fac_compras['CostoUnitario']
fac_compras = fac_compras[['id_fecha', 'CodProducto', 'CodSucursal', 'CodProveedor', 'Unidades', 'CostoUnitario', 'total_compra']]

# -- Fac Ventas --
fac_ventas = df_ventas.copy()
fac_ventas['id_fecha'] = fac_ventas['Fecha'].map(mapa_fechas)
fac_ventas['total_venta'] = fac_ventas['Unidades'] * fac_ventas['PrecioUnitario']
fac_ventas = fac_ventas[['id_fecha', 'CodProducto', 'CodSucursal', 'CodCliente', 'CodVendedor', 'Unidades', 'PrecioUnitario', 'total_venta']]

fac_compras.dropna(subset=['CodProveedor', 'CodProducto', 'CodSucursal'], inplace=True)
fac_ventas.dropna(subset=['CodCliente', 'CodVendedor', 'CodProducto', 'CodSucursal'], inplace=True)

print("Tablas de hechos creadas.")
display(fac_ventas.head(3))
```

# Cargar datos a la BD

```python
print("\n--- 3. Iniciando Carga a MySQL ---")

tablas_etl = {
    'dim_fecha': dim_fecha_final,
    'dim_producto': dim_producto,
    'dim_sucursal': dim_sucursal,
    'dim_cliente': dim_cliente,
    'dim_vendedor': dim_vendedor,
    'dim_proveedor': dim_proveedor,
    'fac_ventas': fac_ventas,
    'fac_compras': fac_compras
}

for nombre, df in tablas_etl.items():
    try:
        # Usamos 'append' para insertar en las tablas ya creadas con SQL
        # index=False es vital para no intentar subir el índice de Pandas
        df.to_sql(name=nombre, con=engine, if_exists='append', index=False)
        print(f"   -> Tabla '{nombre}' cargada ({len(df)} registros).")
    except Exception as e:
        # Si da error de duplicados (IntegrityError), es porque ya corriste el script antes.
        # Puedes ignorarlo o borrar los datos manualmente en MySQL antes de correrlo.
        print(f"   ⚠️ Error cargando '{nombre}': {e}")
        # Sugerencia visual para el usuario
        if "Duplicate entry" in str(e):
            print("      (Los datos ya existían, no se duplicaron).")

print("✅ Carga finalizada.")
```

# Consultas realizadas 

### Consulta 1

```python
# Consulta 1: Total de compras y ventas por año

query1 = """
SELECT 
    f.anio,
    SUM(v.total_venta) as Total_Ventas,
    SUM(c.total_compra) as Total_Compras
FROM dim_fecha f
LEFT JOIN fac_ventas v ON f.id_fecha = v.id_fecha
LEFT JOIN fac_compras c ON f.id_fecha = c.id_fecha
GROUP BY f.anio
ORDER BY f.anio;
"""

try:
    res = pd.read_sql(query1, engine)
    if res.empty:
        print("   (Sin resultados)")
    else:
        print(res.to_string(index=False))
except Exception as e:
    print(f"Error SQL: {e}")
```

### Consulta 2

```python
# Consulta 2: Productos con pérdida (Precio Venta < Costo Compra)

query2 = """
SELECT 
    p.NombreProducto,
    AVG(v.PrecioUnitario) as Precio_Promedio_Venta,
    AVG(c.CostoUnitario) as Costo_Promedio_Compra,
    (AVG(v.PrecioUnitario) - AVG(c.CostoUnitario)) as Margen
FROM dim_producto p
JOIN fac_ventas v ON p.CodProducto = v.CodProducto
JOIN fac_compras c ON p.CodProducto = c.CodProducto
GROUP BY p.NombreProducto
HAVING AVG(v.PrecioUnitario) < AVG(c.CostoUnitario)
ORDER BY Margen ASC;
"""

try:
    res = pd.read_sql(query2, engine)
    if res.empty:
        print("   (Sin resultados)")
    else:
        print(res.to_string(index=False))
except Exception as e:
    print(f"Error SQL: {e}")
```

### Consulta 3

```python
# Consulta 3: Top 5 productos más vendidos por unidades

query3 = """
SELECT 
    p.NombreProducto,
    SUM(v.Unidades) as Total_Unidades_Vendidas
FROM fac_ventas v
JOIN dim_producto p ON v.CodProducto = p.CodProducto
GROUP BY p.NombreProducto
ORDER BY Total_Unidades_Vendidas DESC
LIMIT 5;
"""

try:
    res = pd.read_sql(query3, engine)
    if res.empty:
        print("   (Sin resultados)")
    else:
        print(res.to_string(index=False))
except Exception as e:
    print(f"Error SQL: {e}")
```

### Consulta 4

```python
# Consulta 4: Ingresos por región y año

query4 = """
SELECT 
    s.Region,
    f.anio,
    SUM(v.total_venta) as Ingresos
FROM fac_ventas v
JOIN dim_sucursal s ON v.CodSucursal = s.CodSucursal
JOIN dim_fecha f ON v.id_fecha = f.id_fecha
GROUP BY s.Region, f.anio
ORDER BY s.Region, f.anio;
"""

try:
    res = pd.read_sql(query4, engine)
    if res.empty:
        print("   (Sin resultados)")
    else:
        print(res.to_string(index=False))
except Exception as e:
    print(f"Error SQL: {e}")
```

### Consulta 5

```python
# Consulta: Proveedores con mayor volumen de compras

query5 = """
SELECT 
    p.NombreProveedor,
    SUM(c.Unidades) as Total_Unidades_Compradas,
    SUM(c.total_compra) as Costo_Total_Inversion
FROM fac_compras c
JOIN dim_proveedor p ON c.CodProveedor = p.CodProveedor
GROUP BY p.NombreProveedor
ORDER BY Total_Unidades_Compradas DESC
LIMIT 5;
"""

try:
    res = pd.read_sql(query5, engine)
    if res.empty:
        print("   (Sin resultados)")
    else:
        print(res.to_string(index=False))
except Exception as e:
    print(f"Error SQL: {e}")
```
